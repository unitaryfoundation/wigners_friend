### Branch factor

**Definition (Branch factor)**

Define an arbitrary quantum state

$$
\ket{\Psi_N} = \alpha_1 \ket{\psi_1} + \alpha_2 \ket{\psi_2} + \cdots + \alpha_N \ket{\psi_N}
$$

where $\alpha_k \in \mathbb{C}$ are complex probability amplitudes with the property that $\sum_{k=1}^N |\alpha_k|^2 = 1$ and where $\ket{\psi_k} \in \mathbb{C}^d$ are $d$-dimensional vectors that represent quantum states for all $1 \leq k \leq N$. Then the *branch factor* of $\ket{\Psi_N}$ is defined as

$$
b(\ket{\Psi_N}) = 
\max_{\substack{\alpha_i, \alpha_j \in \mathbb{C} \\ \ket{\psi_i}, \ket{\psi_j} \in \mathbb{C}^d}} 
\alpha_i \alpha_j \left(C_I(\ket{\psi_i}, \ket{\psi_j}) - C_D(\ket{\psi_i}, \ket{\psi_j})\right)
$$

where $C_I(\ket{\psi_i}, \ket{\psi_j})$ is the interference complexity and $C_D(\ket{\psi_i}, \ket{\psi_j})$ is the distinguishability complexity between states $\ket{\psi_i}$ and $\ket{\psi_j}$.

### Examples

#### Generalized GHZ state

As a warm-up, consider the generalized GHZ state on $n$ qubits for some integer $n \geq 1$

$$
\ket{GHZ} = \frac{1}{\sqrt{2}} \left(\ket{0}^{\otimes n} + \ket{1}^{\otimes n}\right).
$$

It follows from [arXiv:2308.04494](https://arxiv.org/pdf/2308.04494) that $C_I(\ket{0}^{\otimes n}, \ket{1}^{\otimes n}) = n$ and $C_D(\ket{0}^{\otimes n}, \ket{1}^{\otimes n}) = 1$, so

$$
b(\ket{GHZ}) = \frac{1}{\sqrt{2}}\frac{1}{\sqrt{2}} \left(C_I(\ket{0}^{\otimes n}, \ket{1}^{\otimes n}) - C_D(\ket{0}^{\otimes n}, \ket{1}^{\otimes n})\right) = \frac{1}{2} \left(n - 1\right)
$$

#### Maximally mixed state

Recall that for some integer $N \geq 1$, the maximally mixed state is defined as

$$
\ket{\Phi_N} = \frac{1}{\sqrt{2^N}} \sum_{i=0}^{2^N-1} \ket{i}.
$$

In this case, the branch factor can be computed as a function of the Hamming distance of the states. 

$$
b(\ket{\Phi_N}) = \max_{\ket{x}, \ket{y}} d_H(\ket{x}, \ket{y})
$$

#### Non-uniform amplitudes

Let $\epsilon \in [0,1]$ and let

$$
\ket{\tau_{\epsilon}} = \sqrt{\frac{1+\epsilon}{2}}\ket{00} + \sqrt{\frac{1-\epsilon}{2}}\ket{11}
$$

Note that when $\epsilon = 0$, the state $\ket{\tau_{0}}$ is the maximally entangled state, so $b(\ket{\tau_{0}}) = 1/2$. Alternatively. when $\epsilon = 1$, the state $\ket{\tau_{1}}$ is a product state, so $b(\ket{\tau_{1}}) = 0$. For $\epsilon \in (0,1)$, we have that $b(\ket{\tau_{\epsilon}}) \in (0, 1/2)$.

**Definition (Mutual coherence)**

Let $S = \{\ket{\psi_1}, \ldots, \ket{\psi_N}\}$ be a collection of vectors of quantum states. Then the *[mutual coherence](https://en.wikipedia.org/wiki/Mutual_coherence_(linear_algebra)* of $S$ is defined as
$$
\mu(S) = \max_{i \neq j} |\langle\psi_i | \psi_j \rangle|.
$$

This quantity provides an indication as to what the maximum overlap is over a given set of vectors. Mutual coherence is used in fields like signal processing and compressed sensing, but may be an interesting metric to use comparatively when it comes to considering the branch factor.
